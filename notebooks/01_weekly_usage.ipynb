```python
import nflreadpy as nfl
import pandas as pd
import matplotlib.pyplot as plt

SEASON = 2024

stats = nfl.load_player_stats([SEASON]).to_pandas()
snaps = nfl.load_snap_counts([SEASON]).to_pandas()
players = nfl.load_players().to_pandas()

print("Player stats:", stats.shape)
print("Snap counts:", snaps.shape)
print("Player metadata:", players.shape)
```

```python
print("STATS")
print(stats.columns.tolist())
display(stats.head())

print("SNAPS")
print(snaps.columns.tolist())
display(snaps.head())

print("PLAYERS")
print(players.columns.tolist())
display(players.head())
```

```python
for name, table in [
    ("STATS", stats),
    ("SNAPS", snaps),
    ("PLAYERS", players),
]:
    print(f"\n{name}:")
    print(table.columns.tolist())
```

```python
stats_reg = stats.loc[stats["season_type"] == "REG"].copy()
snaps_reg = snaps.loc[snaps["game_type"] == "REG"].copy()

team_aliases = {"LA": "LAR"}
stats_reg["team"] = stats_reg["team"].replace(team_aliases)
snaps_reg["team"] = snaps_reg["team"].replace(team_aliases)

stats_reg[["targets", "carries"]] = (
    stats_reg[["targets", "carries"]].fillna(0)
)

team_totals = (
    stats_reg
    .groupby(["season", "week", "team"], as_index=False)
    [["targets", "carries"]]
    .sum()
    .rename(columns={
        "targets": "team_targets",
        "carries": "team_carries",
    })
)

team_totals["team_opportunities"] = (
    team_totals["team_targets"] + team_totals["team_carries"]
)

display(team_totals.head())
```

```python
id_map = (
    players[["pfr_id", "gsis_id"]]
    .dropna(subset=["pfr_id", "gsis_id"])
    .drop_duplicates()
    .rename(columns={
        "pfr_id": "pfr_player_id",
        "gsis_id": "player_id",
    })
)

rb_snaps = snaps_reg.loc[
    (snaps_reg["position"] == "RB")
    & (snaps_reg["offense_snaps"] > 0)
].copy()

rb_snaps = rb_snaps.merge(
    id_map,
    on="pfr_player_id",
    how="left",
    validate="many_to_one",
)

unmatched = rb_snaps.loc[
    rb_snaps["player_id"].isna(),
    ["player", "pfr_player_id"],
].drop_duplicates()

display(unmatched)

assert unmatched.empty, "Some RB IDs did not match. Inspect the table above."
```

```python
join_keys = ["player_id", "season", "week", "team"]

weekly_usage = rb_snaps.merge(
    stats_reg[join_keys + ["targets", "carries"]],
    on=join_keys,
    how="left",
    validate="one_to_one",
    indicator=True,
)

# Review players who participated but have no stats row.
display(
    weekly_usage.loc[
        weekly_usage["_merge"] == "left_only",
        ["player", "season", "week", "team"],
    ].head(20)
)

# An absent stats row can represent a player who had no touches.
# Unexpected names above should be investigated for join problems.
weekly_usage[["targets", "carries"]] = (
    weekly_usage[["targets", "carries"]].fillna(0)
)

weekly_usage = weekly_usage.drop(columns="_merge").merge(
    team_totals,
    on=["season", "week", "team"],
    how="left",
    validate="many_to_one",
)

assert weekly_usage["team_opportunities"].notna().all(), (
    "Some team totals did not match."
)
```

```python
weekly_usage = weekly_usage.rename(columns={
    "player": "player_name",
    "player_id": "gsis_id",
    "offense_pct": "snap_pct",
})

weekly_usage["opportunities"] = (
    weekly_usage["targets"] + weekly_usage["carries"]
)

assert (weekly_usage["team_opportunities"] > 0).all()

weekly_usage["opp_share"] = (
    weekly_usage["opportunities"]
    / weekly_usage["team_opportunities"]
)

weekly_usage = weekly_usage[
    [
        "season", "week", "team", "gsis_id",
        "player_name", "position", "offense_snaps",
        "snap_pct", "targets", "carries",
        "team_opportunities", "opp_share",
    ]
].sort_values(["player_name", "season", "week"])

assert not weekly_usage.duplicated(
    ["gsis_id", "season", "week", "team"]
).any()

for column in ["snap_pct", "opp_share"]:
    assert weekly_usage[column].between(0, 1).all(), (
        f"Check missing or invalid values in {column}"
    )

print("Running-back player-game rows:", len(weekly_usage))
display(weekly_usage.head(10))
```

```python
# Show available player names.
print(sorted(weekly_usage["player_name"].unique()))
```

```python
import matplotlib.pyplot as plt

player_name = "Derrick Henry"  

trend = weekly_usage.loc[
    weekly_usage["player_name"] == player_name
].sort_values(["season", "week"])

assert not trend.empty, "That name wasn't found—check the spelling."

trend.plot(
    x="week",
    y=["snap_pct", "opp_share"],
    marker="o",
    figsize=(10, 5),
    title=f"{player_name}: Weekly Usage",
)

plt.ylabel("Share")
plt.ylim(0, 1)
plt.grid(alpha=0.3)
plt.show()
```
