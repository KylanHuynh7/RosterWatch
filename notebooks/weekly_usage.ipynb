# Data Import

```python
import numpy as np
import pandas as pd
import nflreadpy as nfl
import matplotlib.pyplot as plt

SEASONS = [2022, 2023, 2024, 2025]
POSITIONS = ["RB", "WR", "TE"]

PBP_COLS = [
    "season", "week", "season_type", "posteam", "down", "qtr",
    "half_seconds_remaining", "yardline_100", "pass_attempt", "rush_attempt",
    "qb_kneel", "qb_spike", "receiver_player_id", "rusher_player_id",
]

stats = nfl.load_player_stats(SEASONS).to_pandas()
snaps = nfl.load_snap_counts(SEASONS).to_pandas()
players = nfl.load_players().to_pandas()
ff_ids = nfl.load_ff_playerids().to_pandas()
pbp = nfl.load_pbp(SEASONS).select(PBP_COLS).to_pandas()

print("Player stats:", stats.shape)
print("Snap counts:", snaps.shape)
print("Player metadata:", players.shape)
print("Fantasy football IDs:", ff_ids.shape)
print("Play-by-play:", pbp.shape)
```

# Look at columns 

```python
for name, frame in [("STATS", stats), ("SNAPS", snaps), ("PLAYERS", players)]:
    print(name)
    print(frame.columns.tolist())
    display(frame.head())
```

# Filter and Standardize

```python
TEAM_ALIASES = {"LA": "LAR", "STL": "LAR", "SD": "LAC", "OAK": "LV"}

stats_reg = stats.loc[stats["season_type"] == "REG"].copy()
snaps_reg = snaps.loc[snaps["game_type"] == "REG"].copy()
pbp_reg = pbp.loc[pbp["season_type"] == "REG"].copy()

stats_reg["team"] = stats_reg["team"].replace(TEAM_ALIASES)
snaps_reg["team"] = snaps_reg["team"].replace(TEAM_ALIASES)
pbp_reg["posteam"] = pbp_reg["posteam"].replace(TEAM_ALIASES)

stats_reg[["targets", "carries"]] = stats_reg[["targets", "carries"]].fillna(0)
```

# Add up the team totals

```python
team_totals = (
    stats_reg
    .groupby(["season", "week", "team"], as_index=False)
    [["targets", "carries"]]
    .sum()
    .rename(columns={"targets": "team_targets", "carries": "team_carries"})
)
team_totals["team_opportunities"] = (
    team_totals["team_targets"] + team_totals["team_carries"]
)

display(team_totals.head())
```

# Passing-down and red-zone counts (lowkey a lot)

```python
plays = pbp_reg.loc[
    pbp_reg["posteam"].notna()
    & (pbp_reg["qb_kneel"].fillna(0) == 0)
    & (pbp_reg["qb_spike"].fillna(0) == 0)
].copy()

plays["is_passdown"] = (
    (plays["down"].fillna(0) >= 3)
    | (plays["qtr"].isin([2, 4]) & (plays["half_seconds_remaining"].fillna(9999) <= 120))
).astype(int)
plays["is_redzone"] = (plays["yardline_100"].fillna(100) <= 20).astype(int)

keep = ["season", "week", "posteam", "is_passdown", "is_redzone"]

target_events = (
    plays.loc[(plays["pass_attempt"] == 1) & plays["receiver_player_id"].notna(),
              keep + ["receiver_player_id"]]
    .rename(columns={"posteam": "team", "receiver_player_id": "gsis_id"})
)
carry_events = (
    plays.loc[(plays["rush_attempt"] == 1) & plays["rusher_player_id"].notna(),
              keep + ["rusher_player_id"]]
    .rename(columns={"posteam": "team", "rusher_player_id": "gsis_id"})
)
events = pd.concat([target_events, carry_events], ignore_index=True)

team_sit = (
    events
    .groupby(["season", "week", "team"], as_index=False)
    .agg(
        pbp_team_opportunities=("gsis_id", "size"),
        team_passdown=("is_passdown", "sum"),
        team_redzone=("is_redzone", "sum"),
    )
)
player_sit = (
    events
    .groupby(["season", "week", "team", "gsis_id"], as_index=False)
    .agg(
        player_passdown=("is_passdown", "sum"),
        player_redzone=("is_redzone", "sum"),
    )
)

display(team_sit.head())
```

# Check for totals against pbp totals

```python
opp_check = team_totals.merge(team_sit, on=["season", "week", "team"], how="outer", indicator=True)
print(opp_check["_merge"].value_counts())

opp_check["gap"] = opp_check["team_opportunities"] - opp_check["pbp_team_opportunities"]
print(opp_check["gap"].describe())
print("Share of team-games with any gap:", (opp_check["gap"].fillna(0) != 0).mean().round(3))
```

# Link the player ID's

```python
id_map = (
    players[["pfr_id", "gsis_id"]]
    .dropna(subset=["pfr_id", "gsis_id"])
    .drop_duplicates(subset="pfr_id")
    .rename(columns={"pfr_id": "pfr_player_id", "gsis_id": "player_id"})
)

skill_snaps = snaps_reg.loc[
    snaps_reg["position"].isin(POSITIONS)
    & (snaps_reg["offense_snaps"] > 0)
].copy()

skill_snaps = skill_snaps.merge(
    id_map,
    on="pfr_player_id",
    how="left",
    validate="many_to_one",
)

match_rate = skill_snaps["player_id"].notna().mean()
print(f"ID match rate: {match_rate:.4f}  (prototype: 0.997)")

unmatched = (
    skill_snaps.loc[skill_snaps["player_id"].isna(), ["player", "position", "pfr_player_id"]]
    .value_counts()
    .reset_index(name="games")
)
display(unmatched.head(20))

assert match_rate >= 0.99, "ID match rate dropped below 99%. Inspect the table above."

print("Dropping unmatched player-games:", skill_snaps["player_id"].isna().sum())
skill_snaps = skill_snaps.dropna(subset=["player_id"])
```

# Join snaps to stats and team totals

```python
join_keys = ["player_id", "season", "week", "team"]

weekly_usage = skill_snaps.merge(
    stats_reg[join_keys + ["targets", "carries"]],
    on=join_keys,
    how="left",
    validate="one_to_one",
    indicator=True,
)

display(
    weekly_usage.loc[
        weekly_usage["_merge"] == "left_only",
        ["player", "season", "week", "team"],
    ].head(20)
)
print("Player-games with no stats row:", (weekly_usage["_merge"] == "left_only").sum())

weekly_usage[["targets", "carries"]] = weekly_usage[["targets", "carries"]].fillna(0)

weekly_usage = weekly_usage.drop(columns="_merge").merge(
    team_totals,
    on=["season", "week", "team"],
    how="left",
    validate="many_to_one",
)

assert weekly_usage["team_opportunities"].notna().all(), "Some team totals did not match."
```

# Compute the usage shares

```python
weekly_usage = weekly_usage.rename(columns={
    "player": "player_name",
    "player_id": "gsis_id",
    "offense_pct": "snap_pct",
})

weekly_usage["opportunities"] = weekly_usage["targets"] + weekly_usage["carries"]

# `.where(den > 0)` gives NaN (not 0) when the team had none of that play type.
weekly_usage["target_share"] = weekly_usage["targets"] / weekly_usage["team_targets"].where(weekly_usage["team_targets"] > 0)
weekly_usage["carry_share"] = weekly_usage["carries"] / weekly_usage["team_carries"].where(weekly_usage["team_carries"] > 0)
weekly_usage["opp_share"] = weekly_usage["opportunities"] / weekly_usage["team_opportunities"].where(weekly_usage["team_opportunities"] > 0)

# Situational shares from play-by-play.
weekly_usage = (
    weekly_usage
    .merge(player_sit, on=["season", "week", "team", "gsis_id"], how="left")
    .merge(team_sit[["season", "week", "team", "team_passdown", "team_redzone"]],
           on=["season", "week", "team"], how="left")
)
weekly_usage[["player_passdown", "player_redzone"]] = weekly_usage[["player_passdown", "player_redzone"]].fillna(0)

weekly_usage["passdown_share"] = weekly_usage["player_passdown"] / weekly_usage["team_passdown"].where(weekly_usage["team_passdown"] > 0)
weekly_usage["redzone_share"] = weekly_usage["player_redzone"] / weekly_usage["team_redzone"].where(weekly_usage["team_redzone"] > 0)

weekly_usage = weekly_usage[
    [
        "season", "week", "team", "opponent", "gsis_id", "pfr_player_id",
        "player_name", "position", "offense_snaps",
        "snap_pct", "targets", "carries", "team_opportunities",
        "target_share", "carry_share", "opp_share",
        "passdown_share", "redzone_share",
    ]
].sort_values(["player_name", "season", "week"]).reset_index(drop=True)

display(weekly_usage.head(10))
```

```python
import requests

# 1. Match NFL IDs to Sleeper IDs
sleeper_ids = ff_ids[["gsis_id", "sleeper_id"]].dropna().copy()
sleeper_ids["sleeper_id"] = (
    sleeper_ids["sleeper_id"].astype(str)
    .str.replace(r"\.0$", "", regex=True)
)
sleeper_ids = sleeper_ids.drop_duplicates()

rows = []
weeks = weekly_usage[["season", "week"]].drop_duplicates()

for season, week in weeks.itertuples(index=False, name=None):
    url = f"https://api.sleeper.com/stats/nfl/{int(season)}/{int(week)}"
    response = requests.get(
        url,
        params={"season_type": "regular"},
        timeout=30,
    )
    response.raise_for_status()
    data = response.json()

    if not isinstance(data, list) or not data:
        raise ValueError(f"Unexpected response for {season}, week {week}")

    for record in data:
        rows.append({
            "season": int(season),
            "week": int(week),
            "sleeper_id": str(record["player_id"]),
            "fantasy_points_ppr": (record.get("stats") or {}).get("pts_ppr"),
        })

points = pd.DataFrame(rows)
points["fantasy_points_ppr"] = pd.to_numeric(points["fantasy_points_ppr"])

weekly_usage = (
    weekly_usage
    .drop(columns=["sleeper_id", "fantasy_points_ppr"], errors="ignore")
    .merge(sleeper_ids, on="gsis_id", how="left", validate="many_to_one")
    .merge(
        points,
        on=["season", "week", "sleeper_id"],
        how="left",
        validate="many_to_one",
    )
)

display(
    weekly_usage[
        ["player_name", "season", "week", "fantasy_points_ppr"]
    ].head(10)
)

print("Missing PPR scores:", weekly_usage["fantasy_points_ppr"].isna().sum())
```

```python
id_fixes = {
    "00-0040628": "13210",  # Drake Dabney
    "00-0040406": "12907",  # Quentin Skinner
}

weekly_usage["sleeper_id"] = weekly_usage["sleeper_id"].fillna(
    weekly_usage["gsis_id"].map(id_fixes)
)

weekly_usage = (
    weekly_usage
    .drop(columns="fantasy_points_ppr")
    .merge(
        points,
        on=["season", "week", "sleeper_id"],
        how="left",
        validate="many_to_one",
    )
)

display(
    weekly_usage.loc[
        weekly_usage["gsis_id"].isin(id_fixes)
        & weekly_usage["season"].eq(2025)
        & weekly_usage["week"].eq(18),
        ["player_name", "season", "week", "fantasy_points_ppr"]
    ]
)
```

```python
weekly_usage["fantasy_points_status"] = "available"

weekly_usage.loc[
    weekly_usage["fantasy_points_ppr"].isna(),
    "fantasy_points_status"
] = "score_missing"

weekly_usage.loc[
    weekly_usage["sleeper_id"].isna(),
    "fantasy_points_status"
] = "id_missing"
```

```python
import json
import math
from datetime import datetime, timezone
from pathlib import Path
from urllib.request import Request, urlopen

POSITIONS = {2: "RB", 3: "WR", 4: "TE"}
LIMIT = 5000


def fetch_players(season):
    url = (
        "https://lm-api-reads.fantasy.espn.com/apis/v3/games/ffl/"
        f"seasons/{season}/segments/0/leaguedefaults/3?view=kona_player_info"
    )
    filters = {"players": {
        "limit": LIMIT,
        "sortPercOwned": {"sortPriority": 1, "sortAsc": False},
    }}
    request = Request(url, headers={
        "X-Fantasy-Filter": json.dumps(filters),
        "Accept": "application/json",
        "User-Agent": "RosterWatch/1.0",
    })
    with urlopen(request, timeout=30) as response:
        payload = json.load(response)
    players = payload.get("players") if isinstance(payload, dict) else None
    if not isinstance(players, list) or not players:
        raise ValueError("ESPN returned no players or an unexpected response.")
    if len(players) >= LIMIT:
        raise ValueError("Player limit reached; check pagination before using this snapshot.")
    return players


def numeric(value, field, minimum, maximum):
    if value is None:
        return None
    result = float(value)
    if not math.isfinite(result) or not minimum <= result <= maximum:
        raise ValueError(f"Invalid {field}: {value!r}")
    return result


def make_rows(players, season, collected_at):
    rows, seen = [], set()
    for entry in players:
        player = entry["player"]
        position = POSITIONS.get(player.get("defaultPositionId"))
        if position is None:
            continue
        espn_id = str(player["id"])
        if espn_id in seen:
            raise ValueError(f"Duplicate ESPN ID: {espn_id}")
        seen.add(espn_id)
        ownership = player.get("ownership") or {}
        owned = numeric(ownership.get("percentOwned"), "percentOwned", 0, 100)
        rows.append({
            "season": season,
            "collected_at_utc": collected_at,
            "source": "ESPN",
            "espn_id": espn_id,
            "player_name": player["fullName"],
            "position": position,
            "espn_team_id": player.get("proTeamId"),
            "percentOwned": owned,
            "percentStarted": numeric(ownership.get("percentStarted"), "percentStarted", 0, 100),
            "percentChange": numeric(ownership.get("percentChange"), "percentChange", -100, 100),
            "roster_share": None if owned is None else owned / 100,
            "injury_status": player.get("injuryStatus"),
        })
    if not rows or all(row["percentOwned"] is None for row in rows):
        raise ValueError("No usable RB/WR/TE ownership values returned.")
    return rows

```

```python
OWNERSHIP_SEASON = 2026
espn_players = fetch_players(OWNERSHIP_SEASON)
collected_at = datetime.now(timezone.utc)
espn_ownership = pd.DataFrame(make_rows(
    espn_players, OWNERSHIP_SEASON, collected_at.isoformat()
)).rename(columns={"season": "ownership_season"})

required = {"gsis_id", "espn_id"}
if not required.issubset(ff_ids.columns):
    raise ValueError(f"ff_ids is missing columns: {required - set(ff_ids.columns)}")
espn_ids = ff_ids[["gsis_id", "espn_id"]].dropna().copy()
espn_ids["gsis_id"] = espn_ids["gsis_id"].astype("string").str.strip()
espn_ids["espn_id"] = (
    espn_ids["espn_id"].astype("string").str.strip()
    .str.replace(r"\.0$", "", regex=True)
)
espn_ids = espn_ids.drop_duplicates()
ambiguous = (espn_ids["espn_id"].duplicated(keep=False)
             | espn_ids["gsis_id"].duplicated(keep=False))
ambiguous_espn_ids = espn_ids.loc[ambiguous].copy()
if not ambiguous_espn_ids.empty:
    print("Excluding ambiguous ID mappings (review below):", len(ambiguous_espn_ids))
    display(ambiguous_espn_ids)
espn_ids = espn_ids.loc[~ambiguous].copy()

rostership_snapshot = espn_ownership.merge(
    espn_ids, on="espn_id", how="left", validate="one_to_one"
)
rostership_snapshot["low_rostered"] = (
    rostership_snapshot["percentOwned"].astype("Float64") < 50
)

weekly_usage = weekly_usage.drop(columns="espn_id", errors="ignore").merge(
    espn_ids, on="gsis_id", how="left", validate="many_to_one"
)
print("Snapshot players:", len(rostership_snapshot))
print("Snapshot players without GSIS ID:", rostership_snapshot["gsis_id"].isna().sum())
print("Historical usage ESPN-ID match rate:", weekly_usage["espn_id"].notna().mean())
display(rostership_snapshot[[
    "player_name", "position", "gsis_id", "espn_id", "percentOwned",
    "percentStarted", "percentChange", "low_rostered", "collected_at_utc"
]].head(10))
```

```python
rostership_snapshot.to_csv("espn_ownership.csv", index=False)
print("Saved espn_ownership.csv")
```
