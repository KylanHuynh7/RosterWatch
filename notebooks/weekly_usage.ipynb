# Data Import

```python
import numpy as np
import pandas as pd
import nflreadpy as nfl
import matplotlib.pyplot as plt

SEASONS = [2022, 2023, 2024, 2025]
POSITIONS = ["RB", "WR", "TE"]

PBP_COLS = [
    "season", "week", "season_type", "posteam", "down", "qtr",
    "half_seconds_remaining", "yardline_100", "pass_attempt", "rush_attempt",
    "qb_kneel", "qb_spike", "receiver_player_id", "rusher_player_id",
]

stats = nfl.load_player_stats(SEASONS).to_pandas()
snaps = nfl.load_snap_counts(SEASONS).to_pandas()
players = nfl.load_players().to_pandas()
pbp = nfl.load_pbp(SEASONS).select(PBP_COLS).to_pandas()

print("Player stats:", stats.shape)
print("Snap counts:", snaps.shape)
print("Player metadata:", players.shape)
print("Play-by-play:", pbp.shape)
```

# Look at columns 

```python
for name, frame in [("STATS", stats), ("SNAPS", snaps), ("PLAYERS", players)]:
    print(name)
    print(frame.columns.tolist())
    display(frame.head())
```

# Filter and Standardize

```python
TEAM_ALIASES = {"LA": "LAR", "STL": "LAR", "SD": "LAC", "OAK": "LV"}

stats_reg = stats.loc[stats["season_type"] == "REG"].copy()
snaps_reg = snaps.loc[snaps["game_type"] == "REG"].copy()
pbp_reg = pbp.loc[pbp["season_type"] == "REG"].copy()

stats_reg["team"] = stats_reg["team"].replace(TEAM_ALIASES)
snaps_reg["team"] = snaps_reg["team"].replace(TEAM_ALIASES)
pbp_reg["posteam"] = pbp_reg["posteam"].replace(TEAM_ALIASES)

stats_reg[["targets", "carries"]] = stats_reg[["targets", "carries"]].fillna(0)
```

# Add up the team totals

```python
team_totals = (
    stats_reg
    .groupby(["season", "week", "team"], as_index=False)
    [["targets", "carries"]]
    .sum()
    .rename(columns={"targets": "team_targets", "carries": "team_carries"})
)
team_totals["team_opportunities"] = (
    team_totals["team_targets"] + team_totals["team_carries"]
)

display(team_totals.head())
```

# Passing-down and red-zone counts (lowkey a lot)

```python
plays = pbp_reg.loc[
    pbp_reg["posteam"].notna()
    & (pbp_reg["qb_kneel"].fillna(0) == 0)
    & (pbp_reg["qb_spike"].fillna(0) == 0)
].copy()

plays["is_passdown"] = (
    (plays["down"].fillna(0) >= 3)
    | (plays["qtr"].isin([2, 4]) & (plays["half_seconds_remaining"].fillna(9999) <= 120))
).astype(int)
plays["is_redzone"] = (plays["yardline_100"].fillna(100) <= 20).astype(int)

keep = ["season", "week", "posteam", "is_passdown", "is_redzone"]

target_events = (
    plays.loc[(plays["pass_attempt"] == 1) & plays["receiver_player_id"].notna(),
              keep + ["receiver_player_id"]]
    .rename(columns={"posteam": "team", "receiver_player_id": "gsis_id"})
)
carry_events = (
    plays.loc[(plays["rush_attempt"] == 1) & plays["rusher_player_id"].notna(),
              keep + ["rusher_player_id"]]
    .rename(columns={"posteam": "team", "rusher_player_id": "gsis_id"})
)
events = pd.concat([target_events, carry_events], ignore_index=True)

team_sit = (
    events
    .groupby(["season", "week", "team"], as_index=False)
    .agg(
        pbp_team_opportunities=("gsis_id", "size"),
        team_passdown=("is_passdown", "sum"),
        team_redzone=("is_redzone", "sum"),
    )
)
player_sit = (
    events
    .groupby(["season", "week", "team", "gsis_id"], as_index=False)
    .agg(
        player_passdown=("is_passdown", "sum"),
        player_redzone=("is_redzone", "sum"),
    )
)

display(team_sit.head())
```

# Check for totals against pbp totals

```python
opp_check = team_totals.merge(team_sit, on=["season", "week", "team"], how="outer", indicator=True)
print(opp_check["_merge"].value_counts())

opp_check["gap"] = opp_check["team_opportunities"] - opp_check["pbp_team_opportunities"]
print(opp_check["gap"].describe())
print("Share of team-games with any gap:", (opp_check["gap"].fillna(0) != 0).mean().round(3))
```

# Link the player ID's

```python
id_map = (
    players[["pfr_id", "gsis_id"]]
    .dropna(subset=["pfr_id", "gsis_id"])
    .drop_duplicates(subset="pfr_id")
    .rename(columns={"pfr_id": "pfr_player_id", "gsis_id": "player_id"})
)

skill_snaps = snaps_reg.loc[
    snaps_reg["position"].isin(POSITIONS)
    & (snaps_reg["offense_snaps"] > 0)
].copy()

skill_snaps = skill_snaps.merge(
    id_map,
    on="pfr_player_id",
    how="left",
    validate="many_to_one",
)

match_rate = skill_snaps["player_id"].notna().mean()
print(f"ID match rate: {match_rate:.4f}  (prototype: 0.997)")

unmatched = (
    skill_snaps.loc[skill_snaps["player_id"].isna(), ["player", "position", "pfr_player_id"]]
    .value_counts()
    .reset_index(name="games")
)
display(unmatched.head(20))

assert match_rate >= 0.99, "ID match rate dropped below 99%. Inspect the table above."
```
